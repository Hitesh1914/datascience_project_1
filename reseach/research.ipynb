```python
## Explanation for ensure anotation 

def get_product(x: int, y: int) -> int:
    """
    Function to get product of two numbers
    """
    return x * y
```

```python
get_product(2, 3)
```

```python
get_product(2, "3" )
```

```python
from ensure import ensure_annotations

@ensure_annotations
def get_product(x: int, y: int) -> int:
    """
    Function to get product of two numbers
    """
    return x * y
```

```python
get_product(2, "3" )  # Raises TypeError  Striclty for projects
```
